## Exercise 1: Bayes Rule

*Suppose you are a witness to a nighttime hit-and-run accident involving a taxi in Kiel.*
*All taxi cars in kiel are blue or green.*
*You swear under oath that the taxi was blue.*
*Extensive testing shows that, under the dim lighting conditions, discrimination between blue and green is $75\%$ reliable.*

- (a) *Given your statement as a witness and given that 9 out of 10 Kiel's taxis are green, what is the probability of the taxi being blue?*

---

x : echte Farbe

y : vom Zeugen erkannte Farbe

g : grün

b : blau

P(y = g | x = g) = P(y = b | x = b) = 0.75

---

Gesucht:
P(x = b | y = b) = $\frac {P(y = b | x = b) * P(x = b)} {P(y = b | x = b) * P(x = b) + P(y = b | x = g) * P(x = g)} $

---

P(y = b | x = b) = 0.75

P(y = b | x = g) = 0.25 (Gegenwahrscheinlichkeit)

P(x = b) = 0.1

P(x = g) = 0.9

---

P(x = b | y = b) = $\frac {0.75 * 0.1} {0.75 * 0.1 + 0.25 * 0.9} $

P(x = b | y = b) = 0.25 = 25%

Auch wenn die Wkeit für das korrekte Unterscheiden der Farben bei 75% liegt, beträg die Wkeit dafür, dass das Taxi tatsächlich blau war, nur 25%.

- (b) *Is there a significant change if 7 out of 10 Kiel's taxis are green?*


---

x : echte Farbe

y : vom Zeugen erkannte Farbe

g : grün

b : blau

P(y = g | x = g) = P(y = b | x = b) = 0.75

---

Gesucht:
P(x = b | y = b) = $\frac {P(y = b | x = b) * P(x = b)} {P(y = b | x = b) * P(x = b) + P(y = b | x = g) * P(x = g)} $

---

P(y = b | x = b) = 0.75

P(y = b | x = g) = 0.25 (Gegenwahrscheinlichkeit)

P(x = b) = 0.3

P(x = g) = 0.7

---

P(x = b | y = b) = $\frac {0.75 * 0.3} {0.75 * 0.3 + 0.25 * 0.7} $

P(x = b | y = b) = 0.5625 = 56.25%

Hier liegt die Wkeit dafür, dass das Taxi tatsächlich blau war, nun bei ca. 56%.
Also eine deutliche Veränderung zum vorherigen Durchlauf.

- (c) *Suppose now that there is a second witness who swears that the taxi is green. Unfortunately he is color blind, so he has only a $50\%$ chance of being right. How would this change the estimate from (b)?*

---

x : echte Farbe

y : vom Zeugen 1 erkannte Farbe

z : vom Zeugen 2 (blind) erkannte Farbe

g : grün

b : blau

P(y = g | x = g) = P(y = b | x = b) = 0.75

P(z = g | x = g) = P(z = b | x = b) = 0.5

---

Gesucht:
P(x = b | y = b, z = g) = $\frac {P(y = b, z = g | x = b) * P(x = b)} {P(y = b | x = b) * P(z = g | x = b) * P(x = b) + P(y = b | x = g) * P(z = g | x = g) * P(x = g)} $

y und z haben conditional independence

P(x = b | y = b, z = g) = $\frac {P(y = b | x = b) * P(z = g | x = b) * P(x = b)} {P(y = b | x = b) * P(z = g | x = b) * P(x = b) + P(y = b | x = g) * P(z = g | x = g) * P(x = g)} $

---

P(y = b | x = b) = 0.75

P(y = b | x = g) = 0.25 (Gegenwahrscheinlichkeit)

P(z = g | x = g) = 0.5

P(z = g | x = b) = 0.5 (Gegenwahrscheinlichkeit)

P(x = b) = 0.3

P(x = g) = 0.7

---

P(x = b | y = b, z = g) = $\frac {0.75 * 0.5 * 0.3} {0.75 * 0.5 * 0.3 + 0.25 * 0.5 * 0.7} $

P(x = b | y = b, z = g) = 0.5625 = 56.25%

Hier liegt die Wkeit dafür, dass das Taxi tatsächlich blau war, nun bei ca. 56.25%.
Also keine Veränderung zum vorherigen Durchlauf.
Das kommt daher, da der zusätzliche Zeuge mit 50%igen Wkeit keine zusätzlichen Informationen liefert. Wenn man überall die 0.5 in der Formel rausstreichen würde, hätte man auch wieder den Ansatz aus (b).

## Exercise 2: Bayes Filter

*A vacuum cleaning robot is equipped with a cleaning unit to clean the floor. Furthermore, the robot has a sensor to detect whether the floor is clean or dirty. 
Neither the cleaning unit nor the sensor are perfect. From previous experience you know that the robot succeeds in cleaning a dirty floor with a probability of
\begin{equation*}
	p(x_{t+1} = \mathrm{clean} \;\vert\; x_t = \mathrm{dirty}, u_{t+1} = \mathrm{vacuum}\textrm{-}\mathrm{clean}) = 0.7
\end{equation*}
where $x_{t+1}$ is the state of the floor after having vacuum-cleaned, $u_{t+1}$ is the control command, and $x_t$ is the state of the floor before performing the action.*

*The probability that the sensor indicates that the floor is clean although it is dirty is given by $p(z = \mathrm{clean} \;\vert\; x = \mathrm{dirty}) = 0.3$, and the probability that the sensor correctly detects a clean floor is given by $p(z = \mathrm{clean} \;\vert\; x = \mathrm{clean}) = 0.9$.*

*Unfortunately, you have no knowledge about the current state of the floor. However, after cleaning the floor the sensor of the robot indicates that the floor is clean.*

- (a) *Compute the probability that the floor is still dirty after the robot has vacuum-cleaned it. Use an appropriate prior distribution and justify your choice.*

---

action u = vacuum-cleaned

state x = clean ODER dirty

measurement z = clean ODER dirty

\begin{equation*}
    p1(x_{t+1} = \mathrm{clean} \;\vert\; x_t = \mathrm{dirty}, u_{t+1} = \mathrm{vacuum}\textrm{-}\mathrm{clean}) = 0.7
\end{equation*}

\begin{equation*}
    p2(x_{t+1} = \mathrm{dirty} \;\vert\; x_t = \mathrm{clean}, u_{t+1} = \mathrm{vacuum}\textrm{-}\mathrm{clean}) = 0.0
\end{equation*}

\begin{equation*}
    p3(x_{t+1} = \mathrm{dirty} \;\vert\; x_t = \mathrm{dirty}, u_{t+1} = \mathrm{vacuum}\textrm{-}\mathrm{clean}) = 0.3
\end{equation*}

\begin{equation*}
    p4(x_{t+1} = \mathrm{clean} \;\vert\; x_t = \mathrm{clean}, u_{t+1} = \mathrm{vacuum}\textrm{-}\mathrm{clean}) = 1
\end{equation*}

---

(Eher irrelevant)

Bel($x_{t+1}$ = dirty) = P($x_{t+1}$ = dirty | $z_{t+1}$ = clean, $u_{t+1}$)

η * P(z = clean | $x_{t}$ = dirty, u) * P($x_{t}$ = dirty | z = clean, u) (Bayes)

η * P(z = clean | $x_{t}$ = dirty) * P($x_{t}$ = dirty | z = clean, u) (Markov)

η * P(z = clean | $x_{t}$ = dirty) SUMME P($x_{t}$ = dirty | z = clean, u, $x_{t-1}$ = dirty) * P($x_{t-1}$ = dirty | z = clean, u) $dx_{t-1}$ (Total Probability)

η * P(z = clean | $x_{t}$ = dirty) SUMME P($x_{t}$ = dirty | u, $x_{t-1}$ = dirty) * P($x_{t-1}$ = dirty | z = clean, u) $dx_{t-1}$ (Markov)

η * P(z = clean | $x_{t}$ = dirty) SUMME P($x_{t}$ = dirty | u, $x_{t-1}$ = dirty) * P($x_{t-1}$ = dirty | z = clean, u, $z_{t-1}$ = dirty) $dx_{t-1}$ (Independance Assumption)

η * P(z = clean | $x_{t}$ = dirty) SUMME P($x_{t}$ = dirty | u, $x_{t-1}$ = dirty) * Bel($x_{t-1}$ = dirty) $dx_{t-1}$

---

Da man nichts über den momentanen Zustand des Bodens weiß, wird folgendes als Prior angenommen:

P($x_{t}$ = clean) = 0.5

P($x_{t}$ = dirty) = 0.5

---

Berechnen, wie wahrscheinlich die einzelnen Zustände nach der Reinigung sind:

Zuerst für dirty:

P($x_{t+1}$ = dirty | u) = P($x_{t+1}$ = dirty | $x_{t}$ = clean, u) * P($x_{t}$ = clean) + P($x_{t+1}$ = dirty | $x_{t}$ = dirty, u) * P($x_{t}$ = dirty)

Berechnen:

P($x_{t+1}$ = dirty | u) = 0 * 0.5 + 0.3 * 0.5 = 0.15

Die Gegenwahrscheinlichkeit ergibt dann für clean:

P($x_{t+1}$ = clean | u) = 1 - 0,15 = 0.85

ODER 

P($x_{t+1}$ = clean | u) = P($x_{t+1}$ = clean | $x_{t}$ = clean, u) * P($x_{t}$ = clean) + P($x_{t+1}$ = clean | $x_{t}$ = dirty, u) * P($x_{t}$ = dirty)

P($x_{t+1}$ = clean | u) = 1 * 0.5 + 0.7 * 0.5 = 0.85

---

Nun wird auf den Correction-Part eingegangen (also die Sensormessungen):

P($x_{t+1}$ = dirty | z = clean, u) = $\frac {P(z = clean | x_{t+1} = dirty) * P(x_{t+1} = dirty, u)} {P(z = clean)}$

P(z = clean) = P(z = clean | $x_{t+1}$ = dirty) * P($x_{t+1}$ = dirty, u) + P(z = clean | $x_{t+1}$ = clean) * P($x_{t+1}$ = clean, u)

P(z = clean) = 0.3 * 0.15 + 0.9 * 0.85 = 0.81

Das Ergebnis kann man nun in die vorherige Formel einsetzen:

P($x_{t+1}$ = dirty | z = clean, u) = $\frac {0.3 * 0.15} {0.81}$ = 0.0556 = 5.56%

Somit bleibt der Boden trotz einer positiven Sensormeldung mit einer Wkeit von 5.56% schmutzig.
